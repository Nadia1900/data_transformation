Creating a small neural network in keras that classify different types of wine.

```
#sources
# https://www.tensorflow.org/api_docs/python/tf/keras/utils/to_categorical
# DataCamp - Deep learning with keras
# class materials
```

```
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
import seaborn as sns
import matplotlib.pyplot as plt
```

```
data = pd.read_csv('wines.csv')
data.head()

# The Class column is the classification output variable
```

```
print(data['Class'].values)
```

```
data['Class'].unique()
# the class variable has 3 unique values
```

```
#checking the shape of the data. There are 14 columns and 178 rows. Data is two-dimensional.
data.shape
```

```
X = data.drop(['Class'], axis=1).values #defining X
X.shape # X is 2-dimensional
```

```
y = data['Class'].values #defining y
y.shape # y has 1 dimension
```

```
y=to_categorical(y)
#converting the integer values in y into a binary class matrix and creating dummy variables
```

```
print(y)
#based on the values in the array below, y has 4 classes
```

```
# train_splitting the data and setting the testing size batch to 20%
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```
model = Sequential() #initiaqting a sequential model
model.add(Dense(64, input_shape=(13,), activation='relu')) #adding a dense layer with the input shape corresponding to the number of features
model.add(Dense(64, input_shape=(13,), activation='relu'))
model.add(Dense(64, input_shape=(13,), activation='relu'))
model.add(Dense(4, activation='softmax')) #setting the output layer represents the number of classes
model.compile(optimizer=Adam(0.01), loss='categorical_crossentropy', metrics=['accuracy'])
#compiling the model
```

```
m = model.fit(X_train, y_train, batch_size=10, epochs=15, validation_split=0.2)
#training the model. This batch size, # of epochs gave the most optimal increase in the accuracy rate
```

```
model.predict(X_test) #predicting on the testing data
```

```
model.evaluate(X_test, y_test)
#evaluating the model.
```

```
m.history.keys()
```

```
plt.plot(m.history['loss'])
plt.plot(m.history['val_loss'])

#based on the graph below, it seems that the model is performing well on untrained data.
```
