```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Sample data
data = {
   'Height': [170, 160, 180, 175, 165],
    'Weight': [70, 60, 80, 75, 65],
    'Age': [30, 25, 35, 40, 28],
    'Income': [50000, 60000, 55000, 65000, 52000],
    'Education_Level': [16, 14, 18, 20, 15],
}
```

```python
# Convert to DataFrame
df = pd.DataFrame(data)
```

```python
# Standardize the data
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)
```

```python
# Determine principal components
pca = PCA()
principal_components = pca.fit_transform(scaled_data)
```

```python
# Print the principal components
print("Principal Components:\n", principal_components)
```

```python
# Create a DataFrame for principal components
pc_df = pd.DataFrame(principal_components.T, index=df.columns, columns=[f'PC{i+1}' for i in range(len(principal_components))])

# Round the numbers for easier reading
pc_df_rounded = pc_df.round(4)

print(pc_df_rounded)
```

```python
# Print the explained variance ratio
explained_variance_ratio = pca.explained_variance_ratio_
print("Explained Variance Ratio:\n", explained_variance_ratio)
```

```python
# Create a DataFrame for explained variance ratio
explained_variance_df = pd.DataFrame(explained_variance_ratio, columns=['Explained Variance Ratio'], index=[f'PC{i+1}' for i in range(len(explained_variance_ratio))])

# Round the numbers for easier reading
explained_variance_df_rounded = explained_variance_df.round(8)

print(explained_variance_df_rounded)
```
